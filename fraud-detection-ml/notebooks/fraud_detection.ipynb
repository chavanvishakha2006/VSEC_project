## 2. Exploratory Data Analysis (EDA)

```python
# Plot Class Imbalance Ratio
plt.figure(figsize=(6, 4))
sns.countplot(x='Class', data=df, palette=['#22C55E', '#EF4444'])
plt.title('Class Distribution (0 = Legit, 1 = Fraud)')
plt.yscale('log')
plt.show()
```

```python
# Transaction Amount Distribution
plt.figure(figsize=(10, 4))
sns.histplot(df[df['Class'] == 0]['Amount'], bins=50, color='#22C55E', label='Legit', kde=True, stat='density')
sns.histplot(df[df['Class'] == 1]['Amount'], bins=50, color='#EF4444', label='Fraud', kde=True, stat='density')
plt.xlim(0, 1000)
plt.title('Transaction Amount Density Distribution')
plt.legend()
plt.show()
```
